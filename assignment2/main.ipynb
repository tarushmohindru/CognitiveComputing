# Question 3

```python
import random

rollNo = 1024170231
random.seed(rollNo)
```

```python
L = [random.randint(100, 900) for _ in range(100)]
L
```

```python
odd = [item for item in L if item % 2 != 0]
print("Count: ", len(odd))
print(odd)
```

```python
even = [item for item in L if item % 2 == 0]
print("Count: ", len(even))
print(even)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
prime_numbers = [num for num in L if is_prime(num)]
print("Count: ", len(prime_numbers))
print(prime_numbers)
```

```python
most_freq = max(set(L), key=L.count)
freq = L.count(most_freq)

most_freq, freq
```

# Question 4

```python
rollNo = "1024170231"
digits = [int(digit) for digit in rollNo]

setA = {d * 7 for d in digits}
setB = {d * 9 for d in digits}

setA, setB
```

```python
union = setA.union(setB)
union
```

```python
intersection = setA.intersection(setB)
intersection
```

```python
#difference() returns elements in one set that aren't in the other, whereas symmetric_difference() combines both A-B and B-A into a single set of all unshared elements.
diffAB = setA.difference(setB)
diffBA = setB.difference(setA)
diffAB, diffBA
```

```python
symmDiff = setA.symmetric_difference(setB)
symmDiff
```

```python
is_sub = setA.issubset(setB)
is_super = setB.issuperset(setA)
is_sub, is_super
```

```python
X = int(input("Enter a value to remove from Set A: "))
setA.discard(X)
setA
```

```python
my_dict = {
    "name": "Alex Smith",
    "roll_no": "10210345",
    "branch": "Computer Science",
    "age": 20,
    "city": "Patiala"
}
```

```python
my_dict["location"] = my_dict.pop("city")
my_dict
```

```python
my_dict["cgpa"] = 9.34
my_dict
```

```python
my_dict["age"] += 1
my_dict
```

```python
# pop() returns the value of the removed key, whereas del is a statement that returns nothing (None).

my_dict.pop("branch")
print(my_dict)

my_dict["branch"] = "COPC"

del my_dict["branch"]
print(my_dict)
```

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("DNE")
```

```python
# When two merged dictionaries share keys, the values from the second dictionary (friend_dict) overwrite the first and "win".

friend_dict = {
    "name": "Jordan Lee",
    "roll_no": "10210399",
    "branch": "Electrical Engineering",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
merged_dict
```

```python
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
string_dict
```
