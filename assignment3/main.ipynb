# Question 6

Create a sample dataset (`employees.csv`) containing information about employees in a company.

```python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25",
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5],
})

employees.to_csv("employees.csv", index=False)
employees
```

### a) Shape (number of rows and columns) of the DataFrame

```python
employees.shape
```

### b) Summary of the DataFrame including data types and non-null counts

```python
employees.info()
```

```python
employees.describe()
```

### e) Statistics from the dataset

```python
print("Average salary          :", employees["Salary"].mean())
print("Total bonus paid        :", employees["Bonus"].sum())
print("Youngest employee's age :", employees["Age"].min())
print("Highest rating          :", employees["Rating"].max())
```

### f) Sort the DataFrame by Salary in descending order

```python
employees.sort_values(by="Salary", ascending=False)
```

### g) Categorize employees based on their performance rating

```python
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    return "Average"


employees["Performance_Category"] = employees["Rating"].apply(categorize)
employees[["Name", "Rating", "Performance_Category"]]
```

### h) Identify missing values in the DataFrame

```python
employees.isnull().sum()
```

### i) Rename the Employee_ID column to ID

```python
employees = employees.rename(columns={"Employee_ID": "ID"})
employees.head()
```

### j) Find all employees who have more than 5 years of experience or belong to the IT department

#### i. More than 5 years of experience

```python
employees[employees["Years_of_Experience"] > 5]
```

#### ii. Belong to the IT department

```python
employees[employees["Department"] == "IT"]
```

### k) Add a new column, Tax, which deducts 10% of the Salary

```python
employees["Tax"] = employees["Salary"] * 0.10
employees[["ID", "Name", "Salary", "Tax"]]
```

### l) Save the modified DataFrame to a new CSV file

```python
employees.to_csv("employees_modified.csv", index=False)
pd.read_csv("employees_modified.csv")
```
