# Question 2

Reverse an array and find the most frequent value(s) with their indices.

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
reversed_arr
```

```python
def most_frequent_values_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    highest_count = counts.max()
    most_frequent_values = values[counts == highest_count]

    for value in most_frequent_values:
        indices = np.where(array == value)[0]
        print(f"Value: {value}, Count: {highest_count}, Indices: {indices}")


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print("For x:")
most_frequent_values_and_indices(x)

print("\nFor y:")
most_frequent_values_and_indices(y)
```

# Question 4

Create an evenly spaced 1-D array using `linspace()` and inspect its attributes.

```python
student_name = "Tarush"
tarush = np.linspace(10, 100, 25)
tarush
```

```python
print("Dimensions:", tarush.ndim)
print("Shape     :", tarush.shape)
print("Elements  :", tarush.size)
print("Data type :", tarush.dtype)
print("Bytes     :", tarush.nbytes)
```

```python
tarush_row = tarush.reshape(1, 25)
transpose_with_reshape = tarush_row.reshape(25, 1)
transpose_with_t = tarush_row.T

print("The original 1-D array has shape:", tarush.shape)
print("tarush.T also has shape        :", tarush.T.shape)
print("The reshaped row array can use T:", np.array_equal(transpose_with_reshape, transpose_with_t))
transpose_with_t
```

# Question 5

Create, summarize, reshape, and resize a 2-D array.

```python
ucs420_tarush = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35],
])
ucs420_tarush
```

```python
print("Mean   :", np.mean(ucs420_tarush))
print("Median :", np.median(ucs420_tarush))
print("Maximum:", np.max(ucs420_tarush))
print("Minimum:", np.min(ucs420_tarush))
print("Unique :", np.unique(ucs420_tarush))
```
